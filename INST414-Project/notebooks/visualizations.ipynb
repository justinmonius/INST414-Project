```python
import pandas as pd
from datetime import datetime

# Load the merged dataset
df = pd.read_csv(r"C:\Users\justi\OneDrive\Documents\UMD\Senior\INST414\INST414-Project\INST414-Project\data\processed\merged_players_dataset.csv")

# Convert date_of_birth to datetime
df["date_of_birth"] = pd.to_datetime(df["date_of_birth"], errors="coerce")

# Calculate current age (as of 2025)
current_year = datetime.now().year
df["age"] = current_year - df["date_of_birth"].dt.year

# Filter reasonable ages
df = df[df["age"].between(15, 45)]

age_avg = df.groupby("age", as_index=False)["market_value_eur"].mean()

# Plot
plt.figure(figsize=(9,6))
sns.lineplot(
    data=age_avg,
    x="age",
    y="market_value_eur",
    color="royalblue",
    linewidth=2.5,
    marker="o"
)

plt.title("Average Market Value by Age", fontsize=16)
plt.xlabel("Age (years)", fontsize=13)
plt.ylabel("Average Market Value (€)", fontsize=13)
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="days_missed", y="market_value_eur", data=df, alpha=0.5, color="darkred")

plt.title("Injury Impact: Days Missed vs. Market Value")
plt.xlabel("Total Days Missed")
plt.ylabel("Market Value (€)")
plt.tight_layout()
plt.show()
```
